```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("../steel.csv")

X = data.drop("tensile_strength", axis=1)

plt.figure(figsize=(12, 6))
plt.boxplot(X.values, labels=X.columns)
plt.xticks(rotation=45, ha='right')
plt.title("Feature Scale Differences in steel.csv (Before Scaling)")
plt.ylabel("Raw Feature Values")
plt.tight_layout()
plt.show()
```
