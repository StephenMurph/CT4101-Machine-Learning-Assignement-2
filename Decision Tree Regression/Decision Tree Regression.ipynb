```python
import pandas as pd
import numpy as np

from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import KFold, cross_validate, GridSearchCV
```

```python
data = pd.read_csv("../steel.csv")

X = data.drop("tensile_strength", axis=1)

y = data["tensile_strength"]

X
```

```python

data.corr(numeric_only=True)["tensile_strength"].sort_values()
```

```python
X = data.drop(["tensile_strength",
              "percent_nickel",
              "percent_chromium",
              "percent_manganese"],
                axis=1)

y = data["tensile_strength"]

X
```

```python
kf = KFold(
    n_splits=10,
    shuffle=True,
    random_state=67
)
```

```python
scoring = {
    "mae": "neg_mean_absolute_error",
    "r2": "r2"
}
```

```python
from sklearn.tree import DecisionTreeRegressor

dt_default = DecisionTreeRegressor(random_state=1)

cv_results_default = cross_validate(
    dt_default,
    X, y,
    cv=kf,
    scoring=scoring,
    return_train_score=True
)

default_train_mae = -np.mean(cv_results_default["train_mae"])
default_test_mae  = -np.mean(cv_results_default["test_mae"])
default_train_r2  = np.mean(cv_results_default["train_r2"])
default_test_r2   = np.mean(cv_results_default["test_r2"])

print("=== Decision Trees Default Hyperparameters ===")
print(f"Train MAE: {default_train_mae:.4f}")
print(f"Test  MAE: {default_test_mae:.4f}")
print(f"Train R²:  {default_train_r2:.4f}")
print(f"Test  R²:  {default_test_r2:.4f}")
```

```python
param_grid = {
    "max_depth": [2, 4, 6, 8, 10, None],
    "min_samples_leaf": [1, 2, 4, 6, 8]
}
```

```python
dt = DecisionTreeRegressor(random_state=1)

grid = GridSearchCV(
    estimator=dt,
    param_grid=param_grid,
    scoring="neg_mean_absolute_error",
    cv=kf,
    return_train_score=True
)

grid.fit(X, y)

print("=== Best Hyperparameters (Decision Tree) ===")
print(grid.best_params_)
print(f"Best CV MAE: {-grid.best_score_:.4f}")
```

```python
best_dt = grid.best_estimator_
best_dt
```

```python
cv_results_tuned = cross_validate(
    best_dt,
    X, y,
    cv=kf,
    scoring=scoring,
    return_train_score=True
)

tuned_train_mae = -np.mean(cv_results_tuned["train_mae"])
tuned_test_mae  = -np.mean(cv_results_tuned["test_mae"])
tuned_train_r2  = np.mean(cv_results_tuned["train_r2"])
tuned_test_r2   = np.mean(cv_results_tuned["test_r2"])

print("=== Decision Tree Tuned Hyperparameters ===")
print(f"Train MAE: {tuned_train_mae:.4f}")
print(f"Test  MAE: {tuned_test_mae:.4f}")
print(f"Train R²:  {tuned_train_r2:.4f}")
print(f"Test  R²:  {tuned_test_r2:.4f}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import KFold, cross_validate

max_depth_values = [2, 4, 6, 8, 10, None]
min_leaf_values = [1, 2, 4, 6, 8]

mae_matrix = np.zeros((len(max_depth_values), len(min_leaf_values)))
r2_matrix = np.zeros((len(max_depth_values), len(min_leaf_values)))

for i, depth in enumerate(max_depth_values):
    for j, leaf in enumerate(min_leaf_values):

        model = DecisionTreeRegressor(
            random_state=1,
            max_depth=depth,
            min_samples_leaf=leaf
        )

        scores = cross_validate(
            model,
            X, y,
            cv=kf,
            scoring={"mae": "neg_mean_absolute_error", "r2": "r2"},
            return_train_score=False
        )

        mae = -np.mean(scores["test_mae"])
        r2  =  np.mean(scores["test_r2"])

        mae_matrix[i, j] = mae
        r2_matrix[i, j] = r2

plt.figure(figsize=(8,6))
plt.title("MAE Heatmap ")

cmap = plt.cm.coolwarm

plt.imshow(mae_matrix, cmap=cmap, aspect="auto")

plt.colorbar(label="MAE")

plt.xticks(np.arange(len(min_leaf_values)), min_leaf_values)
plt.yticks(np.arange(len(max_depth_values)), max_depth_values)

for i in range(len(max_depth_values)):
    for j in range(len(min_leaf_values)):
        plt.gca().add_patch(plt.Rectangle((j-0.5, i-0.5), 1, 1,
                                          fill=False, edgecolor='black', linewidth=1))

        plt.text(j, i, f"{mae_matrix[i,j]:.1f}", ha='center', va='center', color='black')

plt.xlabel("min_samples_leaf")
plt.ylabel("max_depth")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8,6))
plt.title("R² Heatmap")

plt.imshow(r2_matrix, cmap=cmap, aspect="auto")
plt.colorbar(label="R²")

plt.xticks(np.arange(len(min_leaf_values)), min_leaf_values)
plt.yticks(np.arange(len(max_depth_values)), max_depth_values)

for i in range(len(max_depth_values)):
    for j in range(len(min_leaf_values)):
        plt.gca().add_patch(plt.Rectangle((j-0.5, i-0.5), 1, 1,
                                          fill=False, edgecolor='black', linewidth=1))
        plt.text(j, i, f"{r2_matrix[i,j]:.2f}", ha='center', va='center', color='black')

plt.xlabel("min_samples_leaf")
plt.ylabel("max_depth")
plt.tight_layout()
plt.show()
```
